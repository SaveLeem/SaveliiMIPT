```python
import csv
import numpy as np
import matplotlib.pyplot as plt

# Читаем csv
col2 = []
col3 = []

with open("data.csv", "r", encoding="cp1251") as file:
    reader = csv.reader(file, delimiter=",")
    next(reader)

    for row in reader:
        col2.append(float(row[1].replace(",", ".")))
        col3.append(float(row[2].replace(",", ".")))

# Рисунок
fig, ax1 = plt.subplots(figsize=(16, 9))

x_measured = col2
y_measured = col3

# Точки
ax1.scatter(x_measured, y_measured, marker='.', color='b', label='Измерения')

# Погрешности
ax1.errorbar(x_measured, y_measured, yerr=0.2, xerr=0.1,
             color='k', linestyle='None', label='Погрешности')

# МНК
k, b = np.polyfit(x_measured, y_measured, 1)
x_line = np.array([min(x_measured), max(x_measured)])
y_line = k * x_line + b
ax1.plot(x_line, y_line, 'r', label=f'МНК: y = {k:.3f}x + {b:.3f}')

ax1.grid()
ax1.set_title("Зависимость Y от X")
ax1.set_xlabel("шреки в штуках")
ax1.set_ylabel("ослы в штуках")
ax1.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(16, 9))
ax1 = fig.add_subplot(311)
ax2 = fig.add_subplot(312)
ax3 = fig.add_subplot(313)

values_small = np.random.normal(0, 10, 100)
values_big = np.random.normal(0, 10, 10000)
values_huge = np.random.normal(0, 10, 10000000)

ax1.hist(values_small, bins=30, density=True, color='steelblue')
ax2.hist(values_big, bins=100, density=True, color='steelblue')
ax3.hist(values_huge, bins=300, density=True, color='steelblue')

ax1.grid()
ax2.grid()
ax3.grid()
ax1.set_title("N = 100")
ax2.set_title("N = 10000")
ax3.set_title("N = 1000000")

ax1.set_xlim(-40, 40)
ax2.set_xlim(-40, 40)
ax1.set_ylim(0, 0.05)
ax2.set_ylim(0, 0.05)
ax3.set_ylim(0, 0.05)
ax3.set_xlim(-40, 40)

plt.tight_layout()
plt.show()
```

```python
import csv
import numpy as np
import matplotlib.pyplot as plt

data = []

with open("iris_data.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file, delimiter=",")
    next(reader)

    for row in reader:
        data.append(float(row[3]))

less_12 = 0
more12_less15 = 0
more_15 = 0

for i in range(len(data)):
    if data[i] <= 1.2:
        less_12 += 1
    elif data[i] > 1.2 and data[i] <= 1.5:
        more12_less15 += 1
    else:
        more_15 += 1

plt.pie([less_12, more12_less15, more_15],
        labels=["less than 1.2", "more12_less15", "more_15"])
plt.title("Very entertaining flower stats")
plt.show()
```
