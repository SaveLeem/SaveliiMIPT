4. Напишите функцию, которая переводит строку введённую с клавиатуры к написанию заглавными буквами если как минимум 3 из 4-х первых символов строки заглавные. Если длина строки меньше 4-х, то проверяется условие, что все символы заглавные. Если это условие не выполнено, то выводится исходная строка.

Пример:

Входные данные:

PYThon3

Выходные данные:

PYTHON3

```python
#считаем, что строка без пробелов, видимо
def change(s):
    if len(s) < 4:
        if not s.isupper():
            return s
        else:
            return s
    else:
        cnt = 0
        for i in range (4):
            if s[i].isupper():
                cnt += 1
        if cnt >= 3:
            return s.upper()
        else:
            return s

s = "QD"
print(change(s))
```

6. С клавиатуры вводится строка. Если длина строки меньше или равна 2, то выводится ASCII код 1-го символа (т. е. с индексом 0), если больше 2, но меньше 10, то выводится сумма ASCII кодов 1-го (c 0-м индексом), серединного (если символов четной число, то берётся символ слева) и последнего. В случае, если длина строки более 10, то выводится ASCII код последнего символа.

Пример:

Входные данные:

ab

Выходные данные:

97

```python
s = input()

if len(s) <= 2:
    print(ord(s[0]))
elif 2 <= len(s) < 10:
    if len(s) % 2 == 0:
        mid = s[len(s)//2 - 1]
    else:
        mid = s[len(s) // 2]
    print(ord(s[0]) + ord(mid) + ord(s[-1]))
elif len(s) >= 10:
    print(ord(s[-1]))
    
```

Дальше клод сделал визуализацию того как меняется площадь, заметаемая планетой, за заданный интервал. Мы ищем угловую скорость радиус вектора планеты, если что. В симуляции со всеми планетами сразу, площадь не будет 100% константной т.к другие планеты вносят вклад. Так же добавил выбор планеты за которой можно следить. Площадь подсчитывается как интеграл квадрата радиус вектора на угловую частоту в данный момент времени пополам (сумма бесконечного количества треугольниоков с бесконечно малым углом при фиксированной вершин). Дальше написаны те части кода, которые были именены и добавлены


```python
def move_space_object(body, dt):
    """Перемещает тело в соответствии с действующей на него силой.

    Параметры:

    **body** — тело, которое нужно переместить.
    """

    ax = body.Fx / body.m
    ay = body.Fy / body.m
    body.Vx += ax * dt
    body.Vy += ay * dt
    body.x += body.Vx * dt
    body.y += body.Vy * dt


def angular_velocity(body, star):
    """Угловая скорость радиус-вектора тела относительно звезды (рад/с).

    omega = (rx * Vy - ry * Vx) / r^2, где (rx, ry) — вектор от звезды к телу,
    (Vx, Vy) — скорость тела относительно звезды.
    """

    rx = body.x - star.x
    ry = body.y - star.y
    vx = body.Vx - star.Vx
    vy = body.Vy - star.Vy
    return (rx * vy - ry * vx) / (rx ** 2 + ry ** 2)


def swept_area(body, star, dt):
    """Площадь, которую радиус-вектор тела заметает за один шаг dt.

    Один шаг численного интегрирования формулы S = 1/2 * integral(r^2 * omega dt):
    dS = 1/2 * r^2 * omega * dt.
    Сумма таких значений за все шаги интервала даёт площадь за интервал.
    """

    r_sqr = (body.x - star.x) ** 2 + (body.y - star.y) ** 2
    return 0.5 * r_sqr * abs(angular_velocity(body, star)) * dt
```
